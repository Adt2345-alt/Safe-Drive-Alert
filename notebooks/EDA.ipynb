## Municipal Pothole Spatial Density & Weather Correlation Analysis
This notebook analyzes defect GPS spatial distribution, weather correlation factors, and traffic volume patterns.

```python
import numpy as np
import pandas as pd
from src.analytics.clustering import HotspotClusterer
from src.analytics.weather_integration import WeatherCorrelator
from src.analytics.traffic_integration import TrafficImpactAnalyzer

# 1. Initialize Analytics Engine
clusterer = HotspotClusterer(eps_meters=50.0, min_samples=5)
weather = WeatherCorrelator()
traffic = TrafficImpactAnalyzer()

print("[*] Analytics modules loaded successfully.")
```

```python
# 2. Perform Hotspot Clustering
cluster_results = clusterer._generate_synthetic_hotspots()
df_hotspots = pd.DataFrame(cluster_results['hotspots'])
display(df_hotspots)
```

```python
# 3. Compute Weather & Traffic Priority Correlation
weather_res = weather.compute_weather_correlation(rainfall_mm=32.0, freeze_cycles=4)
print("Weather Impact:", weather_res['statement'])

enriched_hotspots = traffic.evaluate_hotspot_traffic(cluster_results['hotspots'])
df_traffic = pd.DataFrame(enriched_hotspots)
display(df_traffic[['hotspot_id', 'name', 'vehicles_per_day', 'priority_score', 'dispatch_urgency']])
```
